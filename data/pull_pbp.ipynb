# Pulling play by play data

Pulling 24-25 games that are not already in the directory

```python
from nba_api.stats.endpoints import leaguegamefinder, playbyplayv3
import pandas as pd
import time
import os
```

```python

# -----------------------
# 0. SETTINGS
# -----------------------
SAVE_DIR = "nba_pbp_2024_25"
os.makedirs(SAVE_DIR, exist_ok=True)

# -----------------------
# 1. GET GAME IDS
# -----------------------
games = leaguegamefinder.LeagueGameFinder(
    season_nullable='2024-25',
    season_type_nullable='Regular Season'
)

df_games = games.get_data_frames()[0]
game_ids = df_games['GAME_ID'].drop_duplicates().tolist()

print(f"Total games: {len(game_ids)}")

# -----------------------
# 2. FILTER OUT ALREADY DONE GAMES
# -----------------------
existing_files = set(os.listdir(SAVE_DIR))
done_games = {f.replace(".parquet", "") for f in existing_files}

remaining_games = [gid for gid in game_ids if gid not in done_games]

print(f"Already done: {len(done_games)}")
print(f"Remaining: {len(remaining_games)}")

# -----------------------
# 3. STORAGE
# -----------------------
failed_games = []

# -----------------------
# 4. LOOP (CHECKPOINTED)
# -----------------------
for i, game_id in enumerate(remaining_games):

    print(f"\n[{i+1}/{len(remaining_games)}] Processing {game_id}")

    success = False
    last_error = None

    for attempt in range(3):
        try:
            pbp = playbyplayv3.PlayByPlayV3(
                game_id=game_id,
                timeout=60
            )

            data = pbp.get_dict()
            plays = data['game']['actions']

            df = pd.DataFrame(plays)
            df['gameId'] = game_id

            # -----------------------
            # SAVE CHECKPOINT
            # -----------------------
            path = os.path.join(SAVE_DIR, f"{game_id}.parquet")
            df.to_parquet(path, index=False)

            print(f"✔ Saved {game_id} ({len(df)} rows)")
            success = True
            break

        except Exception as e:
            last_error = str(e)
            print(f"  retry {attempt+1}: {last_error}")
            time.sleep(3)

    if not success:
        print(f"✖ FAILED {game_id}")
        failed_games.append((game_id, last_error))

    # -----------------------
    # RATE LIMIT (1 request / 10 sec)
    # -----------------------
    time.sleep(10)

# -----------------------
# 5. SAVE FAIL LOG
# -----------------------
failed_df = pd.DataFrame(failed_games, columns=["gameId", "error"])

print("\nDONE")
print("Failed games:", len(failed_games))
```

```python
from nba_api.stats.endpoints import leaguegamelog

# -----------------------
# 0. SETTINGS
# -----------------------
SEASONS = [
    "2021-22",
    "2022-23",
    "2023-24",
    "2024-25",
    "2025-26"
]

SAVE_DIR = "nba_gamelog"
os.makedirs(SAVE_DIR, exist_ok=True)

# -----------------------
# 1. FILTER OUT COMPLETED SEASONS
# -----------------------
existing_files = set(os.listdir(SAVE_DIR))
done_seasons = {f.replace(".parquet", "") for f in existing_files}

remaining_seasons = [s for s in SEASONS if s not in done_seasons]

print(f"Already done: {len(done_seasons)}")
print(f"Remaining: {len(remaining_seasons)}")

# -----------------------
# 2. STORAGE
# -----------------------
failed_seasons = []

# -----------------------
# 3. LOOP (CHECKPOINTED)
# -----------------------
for i, season in enumerate(remaining_seasons):

    print(f"\n[{i+1}/{len(remaining_seasons)}] Processing {season}")

    success = False
    last_error = None

    for attempt in range(3):
        try:
            gamelog = leaguegamelog.LeagueGameLog(
                season=season,
                season_type_all_star="Regular Season",
                timeout=60
            )

            df = gamelog.get_data_frames()[0]
            df["SEASON"] = season

            # -----------------------
            # SAVE CHECKPOINT
            # -----------------------
            path = os.path.join(SAVE_DIR, f"gamelog_{season}.parquet")
            df.to_parquet(path, index=False)

            print(f"✔ Saved {season} ({len(df)} rows)")
            success = True
            break

        except Exception as e:
            last_error = str(e)
            print(f"  retry {attempt+1}: {last_error}")
            time.sleep(3)

    if not success:
        print(f"✖ FAILED {season}")
        failed_seasons.append((season, last_error))

    # -----------------------
    # RATE LIMIT (be nice to API)
    # -----------------------
    time.sleep(5)

# -----------------------
# 4. SAVE FAIL LOG
# -----------------------
failed_df = pd.DataFrame(failed_seasons, columns=["season", "error"])

print("\nDONE")
print("Failed seasons:", len(failed_seasons))
```
